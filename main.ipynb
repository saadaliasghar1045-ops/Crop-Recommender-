<h2 style="color: #2471ff;">Project Implementation</h2>

```python
import numpy as np 
import pandas as pd 
import sklearn as sk 
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
import keras as kr 
from keras.models import Sequential
from keras.layers import Input,Dense
from keras.optimizers import Adam
from keras.losses import sparse_categorical_crossentropy
import tensorflow as tf 
import matplotlib.pyplot as plt
```

<h1 style="color: #1c7e8f;">Data Preprocessing</h1>

```python
# Splitting data into X and Y
data_frame = pd.read_csv(r"data\Crop_recommendation.csv")
X = data_frame.drop(columns="label")
Y = data_frame["label"]
```

```python
# Scaling whole train data for PCA 
whole_scaler = sk.preprocessing.StandardScaler()
whole_scaler.fit(X)
X_whole = whole_scaler.transform(X)
# above whole_scaler and X_whole means whole data (all data before splitting into test and train)
```

```python
# Applying PCA 
pca = PCA(n_components=3)
X_pca = pca.fit_transform(X)
pca.explained_variance_ratio_.sum()
```

```python
# visualizing using matplotlib 
%matplotlib tk

fig = plt.figure(figsize=(12, 9))
ax = fig.add_subplot(111, projection="3d")

for crop in Y.unique():
    mask = Y == crop

    ax.scatter(
        X_pca[mask, 0],
        X_pca[mask, 1],
        X_pca[mask, 2],
        label=crop,
        alpha=0.7
    )

ax.set_xlabel("Principal Component 1")
ax.set_ylabel("Principal Component 2")
ax.set_zlabel("Principal Component 3")

ax.set_title("3D PCA Visualization of Crop Recommendation Dataset")

ax.legend(
    bbox_to_anchor=(1.15, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```

```python
# splitting training and testing data 
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.20,random_state=42,stratify=Y,shuffle=True)
```

```python
# Scaling the data 
scaler = sk.preprocessing.StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Encoding the output label 
label_encoder = sk.preprocessing.LabelEncoder()
label_encoder.fit(Y)
Y_train_encoded = label_encoder.transform(Y_train)
Y_test_encoded = label_encoder.transform(Y_test)
for i in range (22):
    print(f"{i} : {label_encoder.inverse_transform([i])}")
```

<h1>Building model Architecture</h1>

```python
tf.random.set_seed(1234)
model = kr.Sequential([
    Input(shape=(7,)),
    Dense(units=128,activation="relu"),
    Dense(units=64,activation="relu"),
    Dense(units=32,activation="relu"),
    Dense(units=16,activation="relu"),
    Dense(units=22,activation="softmax")
])
model.compile(
    optimizer="Adam",
    loss="sparse_categorical_crossentropy",
    metrics=['accuracy']
)
```

```python
model.fit(X_train_scaled,Y_train_encoded,epochs=20,batch_size=8)
```

```python
model.evaluate(X_test_scaled,Y_test_encoded)
```

```python
predictions = model.predict(X_test_scaled)
Y_pred = predictions.argmax(axis=1)
for i in range (Y_pred.shape[0]):
    if Y_pred[i] != Y_test_encoded[i]:
        print(f"Prediction : {label_encoder.inverse_transform([Y_pred[i]])} | Original : {label_encoder.inverse_transform([Y_test_encoded[i]])}")
```
